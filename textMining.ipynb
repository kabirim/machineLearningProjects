```python
import string
import numpy as np
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import tensorflow as tf
import pandas as pd
import nltk
# nltk.download()
from nltk import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer

# Text mining 
# Text Classifications

# Bag of word
# Word embedded

# Tokenizer => StopWord, Stemmer, Puncatuation

text_economics = 'This book is intended for a two-semester course in Economics taught out of the social sciences or business school. ' \
'Principles of Economics aims to teach considerable range and depth of Economic concepts through an approachable style and methodology.' \
' The authors take a three-pronged approach to every chapter: The concept is covered with a “Heads Up” to ward off confusion, ' \
'a real-world application for that concept, and a “You Try It” section to make sure students are staying on top of the concept.'

text_sport = 'Both were chasing their own piece of history. Barcelona were trying to become the first team to claim back-to-back' \
' titles in the Champions League era while Inter were seeking a first European Cup final since 1972 and a first win in the competition since 1965.'

text_innovation = "From Alzheimer's to cancer, earwax can contain valuable indicators to a person's" \
" health. Now scientists are analysing its chemistry in the hope of finding new ways of diagnosing diseases."

# stk = sent_tokenize(text_economics)
text_economics_word_tokenize = word_tokenize(text_economics.lower())
text_sport_word_tokenize = word_tokenize(text_sport.lower())
text_innovation_word_tokenize = word_tokenize(text_innovation.lower())

documents =[text_economics,text_sport,text_innovation]
labels = ['economics','sport','innovation']

stop_words = stopwords.words('english')
puncatuations = list(string.punctuation)
stemmer = PorterStemmer()
filtred_sentences = []
for token in text_economics_word_tokenize:
    if (token not in stop_words) and (token not in puncatuations):
        word = stemmer.stem(token)
        filtred_sentences.append(word)

filtred_sentences
```

```python
countVec = CountVectorizer(lowercase=True)
bow = countVec.fit_transform(documents)

x = bow.toarray()
x.shape
x
```

```python
features = countVec.get_feature_names_out()
features
```

```python
df = pd.DataFrame(x,columns=features)
df
```

```python
# tf*idf

tfidf = TfidfVectorizer()
x_tdidf = tfidf.fit_transform(documents)

x_tdidf = x_tdidf.toarray()

tfidf_features = tfidf.get_feature_names_out()

df_tf_idf = pd.DataFrame(x_tdidf,columns=tfidf_features)
df_tf_idf
```
